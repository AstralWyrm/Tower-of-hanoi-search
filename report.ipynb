## 1.3(c) Relationship and optimality

For $d\geq1$,

$$f(d,4)=2k(d-1)+1.$$

First move the $d-1$ smaller disks onto the two middle pegs in $k(d-1)$ moves. Move the largest disk directly to the final peg once. Reverse the first stage while exchanging the initial and final pegs to put the smaller disks onto the final peg in another $k(d-1)$ moves.

For the lower bound, before the largest disk first moves, its source and destination must both be clear of smaller disks. Ignoring the stationary largest disk and renaming the pegs gives the problem defining $k(d-1)$, so this prefix takes at least that many moves. Reversing the suffix after the largest disk's last move gives the same bound. The largest disk must move at least once. These disjoint stages require at least $2k(d-1)+1$ moves, and the construction attains this bound.

For example, $k(3)=4$, so four disks take $2\times4+1=9$ moves. The formula excludes $d=0$, whose solution length is zero.
